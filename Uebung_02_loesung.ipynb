### Aufgabe

Für die Zahlen von 1 bis `n`, berechnen Sie die Summe der Zahlen und geben Sie das Ergebnis aus. Wählen Sie für `n` eine beliebige natürliche Zahl, zum Beispiel `n` = 5.

<details open style="border: 1px solid #3c763d; background-color: #dff0d8; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #3c763d; cursor: pointer;">✅ Hinweis</summary>
  <p style="color: #333333;">Verwenden Sie das <strong>Akkumulator-Muster</strong>: Initialisieren Sie eine Variable <code>summe = 0</code> vor der Schleife und addieren Sie in jedem Schritt die aktuelle Zahl: <code>summe += zahl</code></p>
</details>

````python
n = 5
summe = 0

for zahl in range(1, n + 1):
    summe += zahl

print("Die Summe der Zahlen von 1 bis " + str(n) + " ist " + str(summe) + ".")
````

### Aufgabe

Erstellen Sie ein Programm, das alle Zahlen von 1 bis `n` überprüft und für jede Zahl ausgibt, ob sie durch 3 oder durch 5 teilbar ist (oder beides). Wählen Sie für `n` eine beliebige Zahl, zum Beispiel `n` = 15.

<details style="border: 1px solid #bce8f1; background-color: #d9edf7; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">💡 Tipp</summary>
  <p style="color: #333333;">Mit dem Modulo-Operator <code>%</code> erhalten Sie den Rest einer Division. Ist der Rest gleich 0, ist die Zahl ohne Rest teilbar: <code>zahl % 3 == 0</code></p>
</details>

````python
n = 15

for zahl in range(1, n + 1):
    if zahl % 3 == 0 and zahl % 5 == 0:
        print(str(zahl) + " ist durch 3 und 5 teilbar.")
    elif zahl % 3 == 0:
        print(str(zahl) + " ist durch 3 teilbar.")
    elif zahl % 5 == 0:
        print(str(zahl) + " ist durch 5 teilbar.")
    else:
        print(str(zahl) + " ist weder durch 3 noch durch 5 teilbar.")
````

### Aufgabe

Erstellen Sie ein Programm, das für die Zahlen von 1 bis `n` die Fakultät (Produkt aller Zahlen von 1 bis `n`) berechnet und ausgibt. Wählen Sie für `n` eine beliebige Zahl, zum Beispiel `n` = 4.

<details open style="border: 1px solid #3c763d; background-color: #dff0d8; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #3c763d; cursor: pointer;">✅ Hinweis</summary>
  <p style="color: #333333;">Die Fakultät ist das Produkt aller ganzen Zahlen von 1 bis <code>n</code>: <code>n! = 1 · 2 · 3 · ... · n</code>.<br>Verwenden Sie das Akkumulator-Muster — aber mit Multiplikation statt Addition. Wichtig: Initialisieren Sie den Akkumulator mit <code>1</code> (nicht mit <code>0</code>).</p>
</details>

````python
n = 4
fakultaet = 1

for zahl in range(1, n + 1):
    fakultaet *= zahl
    print(str(zahl) + "! = " + str(fakultaet))
````

### Aufgabe

Erstellen Sie ein Programm, das die Zahlen von 1 bis `n` ausgibt, jedoch bei den Zahlen, die durch 3 teilbar sind, das Wort "Fizz" und bei den Zahlen, die durch 5 teilbar sind, das Wort "Buzz" ausgibt. Für Zahlen, die sowohl durch 3 als auch durch 5 teilbar sind, soll "FizzBuzz" ausgegeben werden. Wählen Sie für `n` eine beliebige Zahl, zum Beispiel `n` = 20.

<details open style="border: 1px solid #3c763d; background-color: #dff0d8; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #3c763d; cursor: pointer;">✅ Hinweis</summary>
  <p style="color: #333333;">Achten Sie auf die <strong>Reihenfolge der Bedingungen</strong>: Prüfen Sie zuerst, ob die Zahl durch <em>beide</em> Zahlen (3 und 5) teilbar ist — erst danach die Einzelfälle. Sonst wird <code>FizzBuzz</code> nie ausgegeben.</p>
</details>

````python
n = 20

for zahl in range(1, n + 1):
    if zahl % 3 == 0 and zahl % 5 == 0:
        print("FizzBuzz")
    elif zahl % 3 == 0:
        print("Fizz")
    elif zahl % 5 == 0:
        print("Buzz")
    else:
        print(str(zahl))
````

Schreiben Sie ein Skript das die [Fibonacci Folge](https://de.wikipedia.org/wiki/Fibonacci-Folge) bis zur Stelle `n` berechnet. Dabei soll die Funktionsvorschrift `n` mal wiederholt (iterativ) angewendet werden. 

$$f_n = f_{n-1} + f_{n-2}$$

Startwerte sind:
```
f_1 = 1
f_2 = 2
```

Geben Sie die einzelnen Elemente der Folge mit dem print()-Befehl aus.


Wählen Sie für `n` eine beliebige natürliche Zahl, zum Beispiel `n = 10`.

<details open style="border: 1px solid #3c763d; background-color: #dff0d8; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">

````python
n = 10

# Startwerte
f1 = 1
f2 = 2

print("f_1 = " + str(f1))
print("f_2 = " + str(f2))

# Schleife für Position 3 bis n
for i in range(3, n + 1):
    fn = f1 + f2
    print("f_" + str(i) + " = " + str(fn))
    # Werte aktualisieren
    f1 = f2
    f2 = fn
````

# Slicing von strings

In Python gibt es iterierbare Objekte. Dazu gehören zum Beispiel strings, listen und arrays. Aus diesen Objekten lassen sich einzelne Elemente herauslesen indem sie über ihre Position angesprochen werden. Dazu wird an das Objekt eine eckige Klammer angefügt und eine integer Zahl eingetragen. 
Am Beispiel eines string könnte das so aussehen:
```
text = "Toast"

buchstabe = text[2]

print(buchstabe)
```

Es lassen sich auch Untergruppen auslesen indem zwei Zahlen, ein Startwert und ein Endwert, in die eckigen Klammern geschrieben werden, die durch einen Doppeltpukt getrennt sind. 
Dies wird "slicing" genannt und kann wie folgt aussehen:

```
text = "Toast"

buchstaben = text[1:4]

print(buchstaben)
```

Dabei wird der Startwert eingeschlossen, jedoch vor dem Endwert aufgehört.

Werden negative Zahlen eingetragen werden die Elemente von hinten gezählt, zum Beispiel so:

```
text = "Toast"

buchstabe = text[-1]

print(buchstabe)
```

<details open style="border: 1px solid #a94442; background-color: #f2dede; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #a94442; cursor: pointer;">⚠️ Warnung</summary>
  <p style="color: #333333;">Die Indexierung in Python beginnt bei <code>0</code> — das erste Element hat den Index <code>0</code>, das zweite den Index <code>1</code> usw.<br>Beim Slicing <code>text[start:end]</code> ist der <strong>Endwert exklusiv</strong>: <code>text[0:3]</code> liefert die Zeichen an den Positionen 0, 1 und 2 — nicht 3.</p>
</details>

````python
text = "Toast"

buchstabe = text[2]

print(buchstabe)
````

````python
text = "Toast"

buchstaben = text[1:4]

print(buchstaben)
````

````python
text = "Toast"

buchstabe = text[-1]

print(buchstabe)
````

### Aufgabe

Nehmen Sie einen Text als string und geben Sie die einzelnen Wörter aus. Nutzen Sie dazu das string slicing. Als Beispiel können Sie den Aufgabentext verwenden.

<details style="border: 1px solid #bce8f1; background-color: #d9edf7; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">💡 Tipp</summary>
  <p style="color: #333333;">Suchen Sie zuerst die Positionen der Leerzeichen im Text — sie markieren die Wortgrenzen. Extrahieren Sie dann die Wörter mit <code>text[start:end]</code>.</p>

````python
text = "Nehmen Sie einen Text als string und geben Sie die einzelnen Wörter aus."

# Manuelles Slicing: Indexpositionen für die Wörter suchen

wort1 = text[0:6]     # "Nehmen"
wort2 = text[7:10]    # "Sie"
wort3 = text[11:17]   # "einen"
wort4 = text[17:22]   # "Text"
wort5 = text[22:26]   # "als"
wort6 = text[26:33]   # "string"
wort7 = text[33:37]   # "und"
wort8 = text[37:43]   # "geben"
wort9 = text[43:47]   # "Sie"
wort10 = text[47:51]  # "die"
wort11 = text[51:61]  # "einzelnen"
wort12 = text[61:68]  # "Wörter"
wort13 = text[68:72]  # "aus"

# Ausgabe
print(wort1)
print(wort2)
print(wort3)
print(wort4)
print(wort5)
print(wort6)
print(wort7)
print(wort8)
print(wort9)
print(wort10)
print(wort11)
print(wort12)
print(wort13)
````

### Aufgabe

Gegeben ist ein string. Zerlegen Sie ihn in einzelne Wörter und fügen sie diese wieder zusammen so dass der Text einen Sinn ergibt.

```
wilder_text = "Sie Richtigkeit der manuell Ergebnisse. Prüfen die"
```

<details style="border: 1px solid #bce8f1; background-color: #d9edf7; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">💡 Tipp</summary>
  <p style="color: #333333;">Extrahieren Sie zunächst alle Wörter mit String Slicing. Setzen Sie sie anschließend mit <code>+</code> oder einem f-String in der richtigen Reihenfolge wieder zusammen.</p>
</details>

````python
wilder_text = "Sie Richtigkeit der manuell Ergebnisse. Prüfen die"

# Wörter per Slicing extrahieren
wort_pruefen     = wilder_text[40:46]   # "Prüfen"
wort_sie         = wilder_text[0:3]     # "Sie"
wort_manuell     = wilder_text[20:27]   # "manuell"
wort_die         = wilder_text[47:50]   # "die"
wort_richtigkeit = wilder_text[4:15]    # "Richtigkeit"
wort_der         = wilder_text[16:19]   # "der"
wort_ergebnisse  = wilder_text[28:39]   # "Ergebnisse."

# Satz sinnvoll zusammensetzen
sinnvoller_text = (wort_pruefen + " " + wort_sie + " " + wort_manuell + " " +
                   wort_die + " " + wort_richtigkeit + " " + wort_der + " " +
                   wort_ergebnisse)

print(sinnvoller_text)
````

### Aufgabe

Zerlegen Sie einen string in seine Elemente. Dabei sollen loops genutzt werden. 
1. Geben Sie jedes einzelne Element aus.
2. Geben Sie Blöcke mit von je zwei aufeinander folgenden Elementen aus.
3. Geben Sie Blöcke mit von je drei aufeinander folgenden Elementen aus.

Nehmen Sie den ersten Satz dieser Aufgabe als Beispiel für einen entsprechenden Text (string).


<details open style="border: 1px solid #3c763d; background-color: #dff0d8; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #3c763d; cursor: pointer;">✅ Hinweis</summary>
  <p style="color: #333333;">Verwenden Sie <code>range(len(text))</code> für den Index jedes einzelnen Zeichens.<br>Für Zweierblöcke: <code>range(len(text) - 1)</code>, damit der letzte Block noch vollständig ist.<br>Für Dreierblöcke entsprechend: <code>range(len(text) - 2)</code>.</p>
</details>

````python
text = "Zerlegen Sie einen string in seine Elemente."

# 1. Jedes einzelne Zeichen ausgeben
print("Einzelne Zeichen:")
for i in range(len(text)):
    print(text[i])
````

````python
print("\nZweier-Blöcke:")
# 2. Blöcke mit je zwei aufeinander folgenden Zeichen
for i in range(len(text) - 1):
    print(text[i:i+2])
````

````python
print("\nDreier-Blöcke:")
# 3. Blöcke mit je drei aufeinander folgenden Zeichen
for i in range(len(text) - 2):
    print(text[i:i+3])
````

### b)

Der Funktionsaufruf soll so aussehen:

flaechenmoment_rechteck(breite, hoehe)

### c)

Schreiben Sie eine Funktion mit der sich die Durchbiegung $f$ eines einseitig 
eingespannten Balkens berechnen lässt. Als Parameter sollen die Kraft $F$, die 
Stablänge $l$, die Breite $b$ und die Höhe $h$ des Rechteckquerschnitts 
übergeben werden. Das Elastizitätsmodul $E$ ist ein optionaler Parameter 
mit dem Standardwert für Baustahl.
Setzen Sie für das Elastizitätsmodul einen Standardwert für Baustahl 
ein: $E = 210000000000~N/m^2 $

$f = \frac{F \cdot l^3}{3 \cdot E \cdot I_y} $

Der Funktionsaufruf soll so aussehen:
```python
kraft = 7500  # N
stablaenge = 420  # cm
breite = 20  # cm
hoehe = 40  # cm

durchbiegung(kraft, stablaenge, breite, hoehe)
```

Wenn Sie die im Beispiel genannten Zahlen benutzen sollte Ihre Funktion
`8.26875e-06` zurückgeben.

### d)

Lesen Sie die Textdatei mit den Balkendaten ein. Schreiben Sie eine Funktion die eine Liste mit Balkendaten übergeben bekommt, 
für jeden dieser Balken die Durchbiegung ausrechnet und das Ergebnis in einem kleinen Antwortsatz ausgibt.

Lesen sie die Textdatei ein das Sie eine Liste erhalten in der die einzelnen Zeilen als Elemente vorliegen.

Die Zeilen können dann mit der string-Funktion `split()` geteilt werden.

Die Funktion `readlines()` hilft bei Einlesen von Textdateien.


<div style="border: 1px solid #3c763d; background-color: #dff0d8; padding: 10px; border-radius: 5px; margin: 10px 0;">
<strong>✅ Hinweis:</strong> Achten Sie auf konsistente Einheiten! Wenn Kräfte in N und Längen in cm angegeben sind, muss das Elastizitätsmodul ebenfalls in passenden Einheiten vorliegen (N/cm²). Für Baustahl gilt: $E = 21000000~\text{N/cm}^2$. Alternativ können alle Längen in Meter umgerechnet werden.
</div>

````python
# a)
def M_einseitig(kraft, stablaenge):
    M = kraft * stablaenge
    return M

kraft = 7500  # N
stablaenge = 420  # cm

M_einseitig(kraft, stablaenge)
````

````python
#b)
def flaechenmoment_rechteck(breite, hoehe):
    # A * (h**2)/12
    Iy = (breite * hoehe) * (hoehe**2)/12
    return(Iy)
    
breite = 20  # cm
hoehe = 40  # cm

flaechenmoment_rechteck(breite, hoehe)
````

````python
# c)
def durchbiegung(kraft, stablaenge, breite, hoehe, E=210000000000):
    f = (kraft * stablaenge**3) / (3 * flaechenmoment_rechteck(breite, hoehe) * E)
    return f

kraft = 7500  # N
stablaenge = 420  # cm
breite = 20  # cm
hoehe = 40  # cm

durchbiegung(kraft, stablaenge, breite, hoehe)
````

````python
# d) 
datei = open('balken.txt')
datei.readline()  # Kopfzeile überspringen

stablaenge_list = []
kraft_list = []
breite_list = []
hoehe_list = []

for line in datei:
    [laenge, kraft, breite, hoehe] = line.strip().split(',')
    stablaenge_list.append(float(laenge))
    kraft_list.append(float(kraft))
    breite_list.append(float(breite))
    hoehe_list.append(float(hoehe))
````

````python
for i in range(len(kraft_list)):
    print(f'Balken {i+1}: Durchbiegung = {durchbiegung(kraft_list[i], stablaenge_list[i], breite_list[i], hoehe_list[i]):.4e} m')
````

Erstellen Sie y-Werte auf dem Intervall $x∈[-10,10]$ für zwei Funktionen Ihrer Wahl. Ein Beispiel
wäre $y=0.5x^2+4$ und $y=2x^2-3$

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 50)

def f(x):
    return 0.5*x**2 + 4
def f2(x):
    return 2*x**2 -3



plt.plot(x, f(x), label='f(x)')
plt.plot(x, f2(x),'.-',color= 'red', label='f2(x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.title('Schöner Plot')
````

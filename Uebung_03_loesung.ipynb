### Aufgabe 2:

Erstelle eine Liste von 0 bis 5:
- manuell
- mit `range()`
- mit einer `for`-Loop
- mit einer `while`-Loop

<details style="border: 1px solid #bce8f1; background-color: #d9edf7; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">💡 Tipp</summary>
  <p style="color: #333333;">Mit <code>range(start, stop)</code> erzeugen Sie eine Folge von ganzen Zahlen — der <code>stop</code>-Wert ist <strong>exklusiv</strong>. Mit <code>list(range(...))</code> wandeln Sie sie in eine Liste um.</p>
</details>

````python
# manuell:
liste = [0,1,2,3, 4, 5]
print(liste)

# range()
liste = list(range(0,6))
print(liste)

# for loop
liste = []
for i in range(0,6):
    liste.append(i)
print(liste)

# while loop
liste = []
i = 0
while i < 6:
    liste.append(i)
    i +=1
print(liste)
````

### Aufgabe 3:

Erstelle eine Liste von von 15 bis 135 in 15-er Schritten
- mit `range()`
- mit einem `for`-Loop
- mit einem `while`-Loop

<details style="border: 1px solid #bce8f1; background-color: #d9edf7; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">💡 Tipp</summary>
  <p style="color: #333333;">Mit einem dritten Argument legen Sie die Schrittweite fest: <code>range(start, stop, schritt)</code>. Auch hier ist <code>stop</code> exklusiv — für 135 als letzten Wert brauchen Sie <code>stop=136</code>.</p>
</details>

````python
liste = list(range(15,136, 15))
print(liste)

liste = []
for i in range(15,136, 15):
    liste.append(i)
print(liste)


liste = []
i= 15
while i < 136:
    liste.append(i)
    i +=15
print(liste)
````

### Aufgabe 7:

Erzeugen Sie Liste von -10 bis 10 mit der range Funktion und speichern Sie diese in einer Variablen. Benutzen Sie ausschließlich die Methoden der Listenklasse um folgende Manipulationen durchzuführen. 
Geben Sie die Liste nach jeder Änderung aus.

1. Fügen Sie den Wert 11 am Ende der Liste ein.
2. Erweitern Sie die Liste um [12, 13, 14].
3. Entfernen Sie das dritte, sechste und achte Element der Liste.
4. Fügen Sie an der zehnten Stelle den Wert 217 ein.
5. Entfernen Sie alle geraden Zahlen aus der Liste
6. Sortieren Sie die Liste aufsteigend.

Tipp: am Ende sieht die Liste wiefolgt aus: 
Erwarteter Output:
```text
[-9, -6, -5, -3, -1, 1, 3, 5, 7, 9, 11, 13, 217]
``` 

<details open style="border: 1px solid #3c763d; background-color: #dff0d8; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #3c763d; cursor: pointer;">✅ Hinweis</summary>
  <p style="color: #333333;">Wichtige Listenmethoden: <code>.append(x)</code> fügt ein Element ans Ende hinzu, <code>.extend([...])</code> fügt mehrere Elemente hinzu, <code>.pop(i)</code> entfernt das Element an Index <code>i</code>, <code>.insert(i, x)</code> fügt an Position <code>i</code> ein, <code>.remove(x)</code> entfernt den ersten Wert <code>x</code>, <code>.sort()</code> sortiert aufsteigend.</p>
</details>

<details open style="border: 1px solid #a94442; background-color: #f2dede; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #a94442; cursor: pointer;">⚠️ Warnung</summary>
  <p style="color: #333333;">Beim Entfernen von Elementen aus einer Liste <strong>während Sie über dieselbe Liste iterieren</strong> werden Elemente übersprungen — das ist eine klassische Python-Stolperfalle. Mehr dazu in Aufgabe 8.</p>
</details>

````python
liste = list(range(-10, 11))
print(liste)
liste.append(11)
print(liste)
liste.extend([12, 13, 14])
print(liste)
liste.pop(8)
liste.pop(6)
liste.pop(3)
print(liste)
liste.insert(10, 217)
print(liste)
for zahl in liste:
    if zahl % 2 == 0:
        liste.remove(zahl)
print(liste)
liste.sort()
print(liste)
````

### Aufgabe 8: Stolperfalle – Liste während Iteration verändern

Führen Sie folgenden Code aus und beobachten Sie das Ergebnis:

```python
zahlen = [2, 4, 6, 8]
for z in zahlen:
    if z % 2 == 0:
        zahlen.remove(z)
print(zahlen)
```

a) Was wird ausgegeben? Warum werden nicht alle geraden Zahlen entfernt?

b) Korrigieren Sie den Code so, dass wirklich alle geraden Zahlen entfernt werden.

Erwarteter Output (a + b):
```text
Buggy:   [4, 8]
Korrekt: []
```

<details open style="border: 1px solid #a94442; background-color: #f2dede; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #a94442; cursor: pointer;">⚠️ Warnung</summary>
  <p style="color: #333333;">Wenn Sie Elemente aus einer Liste entfernen, <strong>während Sie über dieselbe Liste iterieren</strong>, verschiebt sich der interne Zeiger der Schleife — Elemente werden übersprungen. Deshalb bleibt bei <code>[2, 4, 6, 8]</code> das Ergebnis <code>[4, 8]</code>: nach dem Entfernen von 2 rutscht 4 auf Index 0, der Zeiger ist aber schon bei Index 1 (6).</p>
</details>

````python
# a) Buggy – überspringt Elemente, weil die Liste während der Iteration schrumpft
zahlen = [2, 4, 6, 8]
for z in zahlen:
    if z % 2 == 0:
        zahlen.remove(z)
print("Buggy:  ", zahlen)   # [4, 8]

# b) Korrekt – über eine Kopie iterieren
zahlen = [2, 4, 6, 8]
kopie = zahlen[:]           # echte Kopie via Slicing
for z in kopie:
    if z % 2 == 0:
        zahlen.remove(z)
print("Korrekt:", zahlen)   # []
````

### Aufgabe 9: Elemente korrekt aus einer Liste entfernen

Gegeben ist folgende Liste:

```python
zahlen = [3, 8, 15, 4, 9, 22, 6, 11, 18, 7]
```

Entfernen Sie alle Zahlen, die durch 3 teilbar sind. Erstellen Sie dazu zuerst mit Slicing eine Kopie der Liste und iterieren Sie über diese Kopie.

Erwarteter Output:
```text
[8, 4, 22, 11, 7]
```

<details open style="border: 1px solid #3c763d; background-color: #dff0d8; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #3c763d; cursor: pointer;">✅ Hinweis</summary>
  <p style="color: #333333;"><code>kopie = zahlen[:]</code> erstellt über Slicing eine <strong>echte, unabhängige Kopie</strong> der Liste — Änderungen an <code>zahlen</code> beeinflussen <code>kopie</code> nicht und umgekehrt.</p>
</details>

````python
zahlen = [3, 8, 15, 4, 9, 22, 6, 11, 18, 7]

kopie = zahlen[:]   # echte Kopie via Slicing

for z in kopie:
    if z % 3 == 0:
        zahlen.remove(z)

print(zahlen)
````

# Numerische Integration

<details style="border: 1px solid #bce8f1; background-color: #d9edf7; color: #333333; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">Warum numerisch integrieren?</summary>
  <p style="color: #333333;">
Viele Integrale lassen sich nicht (oder nur mühsam) analytisch lösen – und Messdaten liegen ohnehin nur an diskreten Punkten vor. Numerische Verfahren ersetzen deshalb das <strong>Integral</strong> &#8747; f(x) dx durch eine Summe von Flächenstücken (Ober-/Untersumme, Trapezregel, Monte-Carlo).

Der Preis dafür ist ein <strong>Approximationsfehler</strong>. Wie schnell er mit mehr Stützstellen verschwindet, beschreibt die <strong>Konvergenzordnung</strong>.
</p>
</details>

<details style="border: 1px solid #bce8f1; background-color: #d9edf7; color: #333333; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">Ober- und Untersumme</summary>
  <p style="color: #333333;">
Das Intervall [a,&#8201;b] wird mit n gleichmäßig verteilten <strong>Stützstellen</strong> a = x&#8320; &lt; x&#8321; &lt; &#8230; &lt; x<sub>n-1</sub> = b in n&#8722;1 Teilintervalle der Breite &#916;x = (b&#8722;a)/(n&#8722;1) zerlegt. Auf jedem Teilintervall wird das Maximum O<sub>i</sub> bzw. das Minimum U<sub>i</sub> der Funktion bestimmt. Dann gilt:

<pre>
Untersumme = &#931; &#916;x &#183; U<sub>i</sub>   &#8804;   I   &#8804;   &#931; &#916;x &#183; O<sub>i</sub> = Obersumme
</pre>

Der exakte Integralwert wird also von beiden Seiten <strong>eingeschachtelt</strong>. Mit wachsendem n rücken die Schranken zusammen.
</p>
</details>

<details style="border: 1px solid #bce8f1; background-color: #d9edf7; color: #333333; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">Trapezregel</summary>
  <p style="color: #333333;">
Genauer wird es, wenn die Funktion zwischen den Stützstellen durch <strong>Geraden</strong> angenähert wird – aus den Rechtecken werden Trapeze. Für Funktionswerte y&#8320;, y&#8321;, &#8230;, y<sub>n-1</sub> an gleichmäßig verteilten Stützstellen mit Abstand &#916;x gilt:

<pre>
I &#8776; &#931; &#916;x &#183; (y<sub>i</sub> + y<sub>i+1</sub>) / 2        (Summe über alle n&#8722;1 Teilintervalle)
</pre>

In <code>scipy</code> ist das Verfahren als <code>scipy.integrate.trapezoid(y, x)</code> eingebaut – damit können Sie Ihre eigene Implementierung überprüfen.
</p>
</details>

<details open style="border: 1px solid #3c763d; background-color: #dff0d8; color: #333333; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #3c763d; cursor: pointer;">Konvergenzordnung anschaulich</summary>
  <p style="color: #333333;">
<ul>
<li><strong>1. Ordnung</strong> (z.&nbsp;B. Ober-/Untersumme): 10-mal so viele Stützstellen &nbsp;&#8658;&nbsp; Fehler sinkt etwa auf <strong>1/10</strong>.</li>
<li><strong>2. Ordnung</strong> (Trapezregel): 10-mal so viele Stützstellen &nbsp;&#8658;&nbsp; Fehler sinkt etwa auf <strong>1/100</strong>.</li>
</ul>
Genau das sehen Sie in der Konvergenzstudie in Aufgabenteil B.
</p>
</details>

Ein Dachträger mit Spannweite $L = 8$ m trägt eine ungleichmäßig verteilte Schneelast. Zusammen mit dem Eigengewicht wirkt die Streckenlast

$$ q(x) = 10 + 3 \cdot \sin\left(\frac{\pi x}{L}\right) \quad \text{in kN/m} $$

Für die Auflagerkräfte wird die **Resultierende** der Streckenlast benötigt:

$$ F = \int_0^L q(x)\, dx $$

a) Implementieren Sie $q(x)$ als Python-Funktion und stellen Sie die Lastverteilung mit `matplotlib` dar (Achsenbeschriftungen mit Einheiten und Gitter nicht vergessen – wie in Übung 05).

b) Schreiben Sie eine Funktion `ou_summe(n)`, die die Ober- und die Untersumme des Integrals für `n` gleichmäßig verteilte Stützstellen berechnet und beide Werte zurückgibt. c) Werten Sie `ou_summe` für $n = 5$ und $n = 50$ aus und vergleichen Sie mit dem exakten Wert

d) Warum liegt der exakte Wert immer zwischen Unter- und Obersumme – und was beobachten Sie für wachsendes $n$?

#### Lösungsvorschlag

a)–c) siehe folgende Code-Zelle.

d) Auf jedem Teilintervall gilt $U_i \le q(x) \le O_i$: Die Untersumme **unterschätzt** die Fläche unter der Kurve, die Obersumme **überschätzt** sie – der exakte Wert wird eingeschachtelt. Mit wachsendem $n$ werden die Teilintervalle schmaler, Maximum und Minimum in jedem Teilintervall rücken zusammen, und beide Summen konvergieren gegen $F_{exakt}$. Bei $n = 50$ ist der exakte Wert bereits auf etwa $\pm 0{,}5$ kN eingeschachtelt.

````python
import numpy as np
import matplotlib.pyplot as plt

L = 8.0   # m, Spannweite

def q(x):
    return 10 + 3*np.sin(np.pi*x/L)   # kN/m

# a) Lastverteilung darstellen
x = np.linspace(0, L, 100)
plt.plot(x, q(x))
plt.xlabel("x in m")
plt.ylabel("q(x) in kN/m")
plt.title("Streckenlast auf dem Dachträger")
plt.ylim(0, 14)
plt.grid(True)
plt.show()

# b) Ober- und Untersumme
def ou_summe(n, a=0.0, b=L):
    xi = np.linspace(a, b, n)
    dx = xi[1] - xi[0]
    sum_oben = 0.0
    sum_unten = 0.0
    for i in range(n - 1):
        cx = np.linspace(xi[i], xi[i+1], 50)   # Zwischenpunkte im Teilintervall
        cy = q(cx)
        sum_oben  += dx * np.max(cy)
        sum_unten += dx * np.min(cy)
    return sum_oben, sum_unten

# c) Auswertung und Vergleich mit dem exakten Wert
F_exakt = 10*L + 6*L/np.pi
print(f"Exakte Resultierende: F = {F_exakt:.2f} kN")

for n in [5, 50]:
    oben, unten = ou_summe(n)
    print(f"n = {n:2d}: Untersumme = {unten:6.2f} kN, Obersumme = {oben:6.2f} kN")
````

## Aufgabenteil B – Trapezregel

Die Ober-/Untersumme braucht viele Stützstellen. Die **Trapezregel** aus der Vorlesung ist bei gleichem Aufwand deutlich genauer – das prüfen Sie jetzt an der Streckenlast aus **Aufgabenteil A** nach.

<details style="border: 1px solid #bce8f1; background-color: #d9edf7; color: #333333; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">&#128161; Tipp</summary>
  <p style="color: #333333;">Die Trapezregel lässt sich mit einer <code>for</code>-Schleife über die Teilintervalle umsetzen: In jedem Teilintervall kommt die Trapezfläche <code>dx * (yi[i] + yi[i+1]) / 2</code> hinzu. Für den Vergleich importieren Sie <code>scipy.integrate</code> und rufen <code>scipy.integrate.trapezoid(yi, xi)</code> auf.</p>
</details>

#### Lösungsvorschlag

a), b) siehe folgende Code-Zelle.

c) Der Fehler fällt im halblogarithmischen Diagramm schnell ab: Die Trapezregel konvergiert **quadratisch** (2. Ordnung) – zehnmal so viele Stützstellen bedeuten einen rund hundertmal kleineren Fehler. Die Schwelle von $0{,}01$ kN wird ab $n = 37$ Stützstellen unterschritten (im Diagramm am Schnitt mit der roten Linie ablesbar, im Code per Schleife bestimmt).

````python
import scipy.integrate

# a) eigene Trapezregel
def trapez(yi, dx):
    flaeche = 0.0
    for i in range(len(yi) - 1):
        flaeche += dx * (yi[i] + yi[i+1]) / 2
    return flaeche

# b) Vergleich mit scipy und dem exakten Wert
for n in [5, 50]:
    xi = np.linspace(0, L, n)
    yi = q(xi)
    dx = xi[1] - xi[0]
    print(f"n = {n:2d}: eigene Trapezregel = {trapez(yi, dx):.4f} kN, "
          f"scipy = {scipy.integrate.trapezoid(yi, xi):.4f} kN")

print(f"Exakter Wert: F = {F_exakt:.4f} kN")

# c) Konvergenzstudie
ns = np.arange(2, 51)
fehler = np.zeros(len(ns))
for i, n in enumerate(ns):
    xi = np.linspace(0, L, n)
    fehler[i] = abs(trapez(q(xi), xi[1] - xi[0]) - F_exakt)

plt.plot(ns, fehler)
plt.axhline(y=0.01, color="C3", alpha=0.5)
plt.xlabel("Anzahl der Stützstellen n")
plt.ylabel("Fehler in kN")
plt.title("Konvergenz der Trapezregel")
plt.yscale("log")
plt.grid(True)
plt.show()

for n, f in zip(ns, fehler):
    if f < 0.01:
        print(f"Fehler < 0,01 kN ab n = {n} Stützstellen")
        break
````

### Aufgabe:

Schreiben Sie eine Funktion die eine Liste mit Zahlen entgegen nimmt, jedes Element verdoppelt, und diese Werte in einer neuen Liste sammelt. Diese neue Liste soll dann zurückgegeben werden.

`zahlen = [1, 2, 3, 4, 5, 6, 7, 8, 9]`

Erwarteter Output:

`verdoppeln(liste)`

`[2, 4, 6, 8, 10, 12, 14, 16, 18]`

### Lösung:


<details style="border: 1px solid #bce8f1; background-color: #d9edf7; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">💡 Tipp</summary>
  <p style="color: #333333;">Erstellen Sie vor der Schleife eine leere Liste <code>neue_liste = []</code> und fügen Sie mit <code>.append()</code> in jedem Schritt das verdoppelte Element hinzu.</p>
</details>

````python
zahlen = [1, 2, 3, 4, 5, 6, 7, 8, 9]
````

````python
def verdoppeln(liste):
    neue_liste = list()
    
    for element in liste:
        verdoppelt = element * 2
        neue_liste.append(verdoppelt)
        
    return neue_liste
````

````python
verdoppeln(zahlen)
````

### Aufgabe:

Schreiben Sie eine Funktion mit der sich die Fläche $A$ eines Rechtecks berechnen lässt. Als Parameter sollen die Breite $b$ und die Höhe $h$ des Rechteck übergeben werden.

$A = b \cdot h$



Hilfestellung:

```

hoehe = 40  # cm

rechteck(breite, hoehe)

```

### Lösung:


<details open style="border: 1px solid #3c763d; background-color: #dff0d8; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">

  <p style="color: #333333;">Der Rückgabewert einer Funktion kann direkt weiterverwendet werden: <code>flaeche = rechteck(breite, hoehe)</code>. So lassen sich Berechnungen modular aufbauen.</p>

### Aufgabe:

Schreiben Sie eine Funktion mit der sich die Fläche $A$ eines Kreisrings berechnen lässt. Als Parameter sollen $\pi$, der Innendurchmesser $r$ und der Aussendurchmesser $R$ übergeben werden.
Setzen Sie in der Funktion einen Standardwert für $\pi = 3.1415$ ein

$A = \pi \cdot (R^2~ - ~r^2)$



Hilfestellung:

```
innen = 20  # cm
aussen = 40  # cm

kreisring(innen, aussen, pi)

3769.8

```

### Lösung:


<details open style="border: 1px solid #3c763d; background-color: #dff0d8; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #3c763d; cursor: pointer;">✅ Hinweis</summary>
  <p style="color: #333333;">Standardwerte für Parameter werden direkt in der Funktionsdefinition mit <code>=</code> gesetzt: <code>def kreisring(innen, aussen, pi=3.1415)</code>. Wird beim Aufruf kein Wert für <code>pi</code> übergeben, gilt automatisch <code>3.1415</code>.</p>
</details>

````python
innen = 20  # cm
aussen = 40  # cm

def kreisring(innen, aussen, pi=3.1415):
    flaeche = pi * (aussen**2 - innen**2)
    return flaeche
````

````python
kreisring(innen, aussen)
````

### Aufgabe:

Schreiben Sie ein Programm, welches einen Geldbetrag von Euro ("Euro") in Pfund Sterling ("Pfund") und vice versa umrechnen kann. Dabei soll bei der Eingabe mit angegeben werden, in welcher Währung sich der Geldbetrag momentan befindet. Die Umrechnung soll als string ausgegeben werden.

Erwarteter Output:

`waehrungsumrechnung(geldbetrag, momentare_waehrung)`

`"10.00 Euro entsprechen 8.70 Pfund Sterling."`

### Lösung:


<details style="border: 1px solid #bce8f1; background-color: #d9edf7; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">💡 Tipp</summary>
  <p style="color: #333333;">Mit <code>:.2f</code> in einem f-String wird eine Zahl auf zwei Nachkommastellen formatiert: <code>f"{wert:.2f}"</code> gibt z.B. <code>10.00</code> statt <code>10.0</code> aus.</p>
</details>

````python
def waehrungsumrechnung(geldbetrag, momentare_waehrung):
    if momentare_waehrung == "Euro":
        geldbetrag_euro = geldbetrag
        geldbetrag_pfund = 0.87 * geldbetrag_euro
        return f'{geldbetrag_euro:.2f} Euro entsprechen {geldbetrag_pfund:.2f} Pfund Sterling.'
    if momentare_waehrung == "Pfund":
        geldbetrag_pfund = geldbetrag
        geldbetrag_euro = 1.15 * geldbetrag_pfund
        return f'{geldbetrag_pfund:.2f} Pfund Sterling entsprechen {geldbetrag_euro:.2f} Euro.'
    else:
        return "Der Geldbetrag kann leider nicht in diese Währung umgerechnet werden."
````

````python
print(waehrungsumrechnung(10, "Euro"))
print(waehrungsumrechnung(8.7, "Pfund"))
print(waehrungsumrechnung(23.99, "Pfund"))
````

### Aufgabe:

Schreiben Sie eine Funktion mit der sich das Axiales Flächenmoment 2. Grades um die y-Achse $I_y$ eines Rechtecks berechnen lässt. Nehmen Sie and das der Querschnitt der eines Rechteck ist. Als Parameter sollen die Breite $b$ und die Höhe $h$ des Rechteck übergeben werden.

$I_y = \frac{b \cdot h^3}{12} = A~\frac{h^2}{12}$



Hilfestellung:

```
breite = 20  # cm
hoehe = 40  # cm

flaechenmoment_rechteck(breite, hoehe)

106666.66666666667

```

### Lösung:


<details style="border: 1px solid #bce8f1; background-color: #d9edf7; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">💡 Tipp</summary>
  <p style="color: #333333;">Nutzen Sie die bereits definierte <code>rechteck()</code>-Funktion innerhalb dieser Funktion, um die Fläche zu berechnen — statt die Formel zu wiederholen.</p>

### Aufgabe:

Schreiben Sie eine Funktion mit der sich die Durchbiegung $f$ eines einseitig eingespannten Balkens berechnen lässt. Als Parameter sollen die Kraft $F$, die Stablänge $l$, das Elastizitätsmodul $E$ und die Flächenmoment 2. Grades $I$ übergeben werden.
Setzen Sie für das Elastizitätsmodul einen Standardwert für Baustahl ein: $E = 210000000000~N/m^2 $


$f = \frac{F \cdot l^3}{3 \cdot E \cdot I_y} $



Hilfestellung:

```
kraft = 7500  # N
stablaenge = 420  # cm
breite = 20  # cm
hoehe = 40  # cm

durchbiegung(kraft, stablaenge, breite, hoehe)

8.26875e-06

```

### Lösung:


<details open style="border: 1px solid #3c763d; background-color: #dff0d8; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #3c763d; cursor: pointer;">✅ Hinweis</summary>
  <p style="color: #333333;"><code>I_y</code> muss vor dem Aufruf dieser Funktion berechnet werden, z.B. mit <code>I_y = flaechenmoment_rechteck(breite, hoehe)</code>. Das Ergebnis wird dann als Argument übergeben.</p>
</details>

````python
kraft = 7500  # N
stablaenge = 420  # cm
breite = 20  # cm
hoehe = 40  # cm

def durchbiegung(kraft, stablaenge, I_y, E=210000000000):
    f = (kraft * stablaenge**3) / (3 * E * I_y)
    return f
````

````python
I_y = flaechenmoment_rechteck(breite, hoehe)

durchbiegung(kraft, stablaenge, I_y)
````

### Aufgabe:

Schreiben Sie eine Funktion die eine Textdatei mit Informationen zu verschiedenen Balken einliest, für jeden dieser Balken die Durchbiegung ausrechnet und das Ergebnis in einem kleinen Antwortsatz ausgibt.

Lesen sie die Textdatei ein das Sie eine Liste erhalten in der die einzelnen Zeilen als Elemente vorliegen.

Die Zeilen können dann mit der string-Funktion `split()` geteilt werden.

Die Funktion `readlines()` hilft bei Einlesen von Textdateien.


Hilfestellung:

```
# String teilen.
zeile = "a,b,c,d"
teile = zeile.split(",")
print(teile)
["a", "b", "c", "d"]


```

### Lösung:


<details open style="border: 1px solid #3c763d; background-color: #dff0d8; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #3c763d; cursor: pointer;">✅ Hinweis</summary>
  <p style="color: #333333;"><code>open("datei.txt")</code> öffnet eine Textdatei. <code>readlines()</code> liest alle Zeilen als Liste ein. Mit <code>.strip()</code> entfernen Sie Zeilenumbrüche (<code>\n</code>) und Leerzeichen. Mit <code>.split(",")</code> teilen Sie eine Zeile an Kommas auf.</p>
</details>

````python
with open('balken.txt') as f:
    balken = f.readlines()

    
    
def balken_berechnen(balken):
    
    for b in balken:
        teile = b.strip().split(",")
        
        kraft = int(teile[1])
        stablaenge = int(teile[0])
        breite = int(teile[2])
        hoehe = int(teile[3])
        
        I_y = flaechenmoment_rechteck(breite, hoehe)

        db = durchbiegung(kraft, stablaenge, I_y)
        print(f"Durchbiegung: {db} mm")

    
balken_berechnen(balken[1:-1])
````

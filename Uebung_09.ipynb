# Algorithmen und Flussdiagramme

<details style="border: 1px solid #bce8f1; background-color: #d9edf7; color: #333333; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">Was ist ein Algorithmus?</summary>
  <p style="color: #333333;">
Ein <strong>Algorithmus</strong> ist eine eindeutige, aus endlich vielen Schritten bestehende Handlungsvorschrift zur Lösung eines Problems. Auch außerhalb der Informatik begegnen Ihnen Algorithmen ständig – z.&nbsp;B. als Montageanleitung, Prüfvorschrift oder Bauablaufplan.

Wichtige Eigenschaften eines Algorithmus:

<ul>
<li><strong>Eindeutigkeit:</strong> Jeder Schritt ist unmissverständlich beschrieben.</li>
<li><strong>Ausführbarkeit:</strong> Jeder Schritt ist tatsächlich durchführbar.</li>
<li><strong>Endlichkeit (Terminierung):</strong> Der Algorithmus endet nach endlich vielen Schritten.</li>
<li><strong>Determinismus:</strong> Gleiche Eingaben führen immer zum gleichen Ergebnis.</li>
<li><strong>Allgemeinheit:</strong> Der Algorithmus löst nicht nur ein einzelnes Beispiel, sondern eine ganze Klasse von Problemen.</li>
</ul>
</p>
</details>

<details style="border: 1px solid #bce8f1; background-color: #d9edf7; color: #333333; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">Flussdiagramm-Symbole (Notation aus der Vorlesung)</summary>
  <p style="color: #333333;">
Ein <strong>Flussdiagramm</strong> stellt einen Algorithmus grafisch dar. Wir verwenden dieselben Symbole wie im Vorlesungsskript:

<img src="uebung09_legende.png" alt="Flussdiagramm" style="max-width:100%; height:auto;">

<ul>
<li><strong>Ein-/Ausgabe (rot):</strong> Daten, die in den Algorithmus hineingehen bzw. ihn verlassen – sie markieren Anfang und Ende.</li>
<li><strong>Anweisung (grün):</strong> eine Operation oder Berechnung, z.&nbsp;B. &#8222;Berechne Fläche&#8220;.</li>
<li><strong>Entscheidung (blaue Raute):</strong> Abfrage einer Bedingung mit zwei Ausgängen (ja/nein bzw. wahr/falsch).</li>
<li><strong>Schleife (orange Ellipse):</strong> Schleifenkopf, z.&nbsp;B. &#8222;für alle a in A&#8220; oder &#8222;solange n &gt; 0 wiederhole&#8220;. Ein Ausgang führt in den Schleifenkörper (mit Rücksprunglinie zurück zur Ellipse), der andere (&#8222;fertig&#8220;) verlässt die Schleife.</li>
<li><strong>Grauer Kasten:</strong> fasst den eigentlichen Algorithmus zusammen – Ein- und Ausgabe liegen außerhalb.</li>
</ul>
</p>
</details>

<details style="border: 1px solid #bce8f1; background-color: #d9edf7; color: #333333; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">Beispiel: Flussdiagramm &#8594; Pseudocode &#8594; Python</summary>
  <p style="color: #333333;">
Auf der Baustelle wird die Betondeckung c<sub>vorh</sub> der Bewehrung gemessen und mit der geforderten Mindestbetondeckung c<sub>min</sub> verglichen. Der Prüf-Algorithmus als Flussdiagramm:

<img src="uebung09_beispiel_betondeckung.png" alt="Flussdiagramm" style="max-width:100%; height:auto;">

Derselbe Algorithmus als <strong>Pseudocode</strong>:

<pre>
BEGINN
    LIES c_vorh, c_min
    WENN c_vorh &#8805; c_min DANN
        GIB "Deckung ausreichend" AUS
    SONST
        GIB "Nachbesserung nötig" AUS
ENDE
</pre>

Und als <strong>Python-Programm</strong>:

<pre>
c_vorh = 35   # mm, gemessene Betondeckung
c_min  = 30   # mm, geforderte Mindestbetondeckung

if c_vorh >= c_min:
    print("Deckung ausreichend")
else:
    print("Nachbesserung nötig")
</pre>

Alle drei Darstellungen beschreiben exakt denselben Algorithmus – nur die Notation unterscheidet sich.
</p>
</details>

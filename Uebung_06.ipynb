In der folgenden Zelle werden verschiedene Datenserien erzeugt. Führen Sie diese Zelle einfach aus um die Datenserien in den speicher zu laden.

- a) scatter plot (einzelne Punkte) erstellen und sich die Daten zunächst anschauen
- b) die Daten durch ein Polynom n-ten Gerades annähern -- wählen Sie einen sinnvollen Polynomgrad
- c) plotten Sie das Polynom, in schwarzer Linienfarbe, zusammen mit den jeweilgen Daten
- d) achten Sie auf die Beschriftung der Plots




```python
import numpy as np

# Aufgabe a
# -------------
xmin = 0
xmax = 5
n_a = 25

x_a = np.linspace(xmin, xmax, n_a) + 0.2*(2 * np.random.random(n_a) - 1)
y_a = 2*x_a + 0.5 + 2*(2 * np.random.random(n_a) - 1)


# Aufgabe b
# -------------
xmin = -2
xmax = 4
n_b = 25

x_b = np.linspace(xmin, xmax, n_b) + 0.1*(2 * np.random.random(n_b) - 1)
y_b = x_b**2 - 2*x_b - 3 + 1.5*(2 * np.random.random(n_b) - 1)


# Aufgabe c
# -------------
xmin = -3
xmax = 6
n_c = 25

x_c = np.linspace(xmin, xmax, n_c) + 0.15*(2 * np.random.random(n_c) - 1)
y_c = (x_c + 2) * (x_c - 1) * (x_c - 5) + 5*(2 * np.random.random(n_c) - 1)


# Aufgabe d
# -------------
xmin = 0
xmax = 3
n_d = 25

x_d = np.linspace(xmin, xmax, n_d) + 0.05*(2 * np.random.random(n_d) - 1)
y_d = np.exp(x_d) + 1*(2 * np.random.random(n_d) - 1)


# Aufgabe e
# -------------
xmin = 1
xmax = 10
n_e = 25

x_e = np.linspace(xmin, xmax, n_e) + 0.2*(2 * np.random.random(n_e) - 1)
y_e = np.log(x_e) + 0.2*(2 * np.random.random(n_e) - 1)


# Aufgabe f
# -------------
xmin = 0
xmax = 2 * np.pi
n_f = 25

x_f = np.linspace(xmin, xmax, n_f) + 0.1*(2 * np.random.random(n_f) - 1)
y_f = np.cos(x_f) + 0.15*(2 * np.random.random(n_f) - 1)
```

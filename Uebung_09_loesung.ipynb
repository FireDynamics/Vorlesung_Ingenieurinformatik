<details style="border: 1px solid #bce8f1; background-color: #d9edf7; color: #333333; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">Was ist ein Algorithmus?</summary>
  <p style="color: #333333;">
Ein <strong>Algorithmus</strong> ist eine eindeutige, aus endlich vielen Schritten bestehende Handlungsvorschrift zur Lösung eines Problems. Auch außerhalb der Informatik begegnen Ihnen Algorithmen ständig – z.&nbsp;B. als Montageanleitung, Prüfvorschrift oder Bauablaufplan.

Wichtige Eigenschaften eines Algorithmus:

<ul>
<li><strong>Eindeutigkeit:</strong> Jeder Schritt ist unmissverständlich beschrieben.</li>
<li><strong>Ausführbarkeit:</strong> Jeder Schritt ist tatsächlich durchführbar.</li>
<li><strong>Endlichkeit (Terminierung):</strong> Der Algorithmus endet nach endlich vielen Schritten.</li>
<li><strong>Determinismus:</strong> Gleiche Eingaben führen immer zum gleichen Ergebnis.</li>
<li><strong>Allgemeinheit:</strong> Der Algorithmus löst nicht nur ein einzelnes Beispiel, sondern eine ganze Klasse von Problemen.</li>
</ul>
</p>
</details>

#### Lösungsvorschlag

| Serie | $x_1$ | $x_2$ | $x_3$ | $m$ | $x_{min}$ | Ergebnis |
| :--: | :--: | :--: | :--: | :--: | :--: | :--: |

- Serie 1: $m = 33 \ge 29$ ✓ und $x_{min} = 31 \ge 21$ ✓ → konform
- Serie 2: $m = 29 \ge 29$ ✓ (Grenzfall!) und $x_{min} = 26 \ge 21$ ✓ → konform
- Serie 3: $m = 30 \ge 29$ ✓, aber $x_{min} = 19 < 21$ → nicht konform (zweite Raute)
- Serie 4: $m = 27 < 29$ → nicht konform (**erste Raute**)

Für Serie 4 endet die Prüfung bereits an der **ersten Raute** ($m \ge f_{ck}+4$?). Der Zweig mit $x_{min}$ wird gar nicht mehr durchlaufen – $x_{min}$ wird also nach diesem Algorithmus **nicht berechnet**.

## Aufgabenteil B – Vom Flussdiagramm zum Programm: Betonprüfung in Python

Implementieren Sie nun den Algorithmus aus **Aufgabenteil A** als Python-Funktion:

```python
def pruefe_beton(werte, fck):
    ...
```

Die Funktion erhält eine Liste `werte` mit drei Druckfestigkeiten und die charakteristische Festigkeit `fck` und gibt den Text `"konform"` oder `"nicht konform"` zurück.

Prüfen Sie anschließend mit einer Schleife alle vier Serien aus Aufgabenteil A und vergleichen Sie mit Ihrer Handrechnung.

Erwarteter Output:
```text
Serie 1 : konform
Serie 2 : konform
Serie 3 : nicht konform
Serie 4 : nicht konform
```

<details style="border: 1px solid #bce8f1; background-color: #d9edf7; color: #333333; padding: 10px 15px; border-radius: 4px; margin: 1em 0;">
  <summary style="font-weight: bold; color: #31708f; cursor: pointer;">&#128161; Tipp</summary>
  <p style="color: #333333;">Nützliche eingebaute Funktionen: <code>sum(liste)</code>, <code>len(liste)</code> und <code>min(liste)</code>. Die beiden Rauten des Flussdiagramms können Sie entweder als <strong>verschachtelte</strong> <code>if</code>-Abfragen umsetzen oder mit <code>and</code> zu einer Bedingung kombinieren.</p>
</details>

#### Lösungsvorschlag

Siehe folgende Code-Zelle.

````python
def pruefe_beton(werte, fck):
    m = sum(werte) / len(werte)
    x_min = min(werte)
    if m >= fck + 4:
        if x_min >= fck - 4:
            return "konform"
        else:
            return "nicht konform"
    else:
        return "nicht konform"

# alternativ, mit einer kombinierten Bedingung:
# if m >= fck + 4 and x_min >= fck - 4: ...

serien = [[33, 35, 31],
          [26, 28, 33],
          [35, 36, 19],
          [27, 26, 28]]
fck = 25

for i in range(len(serien)):
    print("Serie", i + 1, ":", pruefe_beton(serien[i], fck))
````

b) Die Umsetzung in Python steht in der folgenden Code-Zelle.

c) Für $z = 20 > p = 15$ **steigt** das Wasservolumen in jeder Stunde – die Bedingung `V > 0` bleibt für immer wahr und das Programm gerät in eine **Endlosschleife**. Verletzt ist die Eigenschaft der **Endlichkeit (Terminierung)**: Der Ablauf endet nicht mehr nach endlich vielen Schritten.

Absichern kann man das Programm z. B. so:
- **Vorabprüfung:** vor der Schleife testen, ob `z >= p`, und in diesem Fall nur eine Meldung ausgeben (`if z >= p: print("Pumpe zu schwach!")`), oder
- **Notbremse:** eine maximale Stundenzahl festlegen, nach der die Schleife abgebrochen wird.

````python
V = 120.0   # m³ Wasser in der Baugrube
p = 15.0    # m³/h Pumpleistung
z = 5.0     # m³/h Zufluss

t = 0
while V > 0:
    V = V - (p - z)
    t = t + 1

print("Die Baugrube ist nach", t, "Stunden leer.")
````

#### Lösungsvorschlag

a) Siehe folgende Code-Zelle. Die Listen `stunden` und `wasserstand` starten mit dem Anfangszustand (Stunde 0, 120 m³), damit die Kurve am richtigen Punkt beginnt.

b) Mit $V = 125$ m³ würde in der letzten Stunde rechnerisch ein **negativer Wasserstand** ($-5$ m³) gespeichert – die Kurve würde unter die Null-Linie fallen, was physikalisch keinen Sinn ergibt (die Pumpe fördert dann nur noch das Restwasser). Die `if`-Abfrage in der Schleife setzt solche Werte auf 0.

````python
import matplotlib.pyplot as plt

V = 120.0   # m³ Wasser in der Baugrube
p = 15.0    # m³/h Pumpleistung
z = 5.0     # m³/h Zufluss

t = 0
stunden = [0]
wasserstand = [V]

while V > 0:
    V = V - (p - z)
    if V < 0:
        V = 0.0
    t = t + 1
    stunden.append(t)
    wasserstand.append(V)

print("Die Baugrube ist nach", t, "Stunden leer.")

plt.plot(stunden, wasserstand, marker="o")
plt.xlabel("Zeit in h")
plt.ylabel("Wasservolumen in m³")
plt.title("Wasserhaltung in der Baugrube")
plt.grid(True)
plt.show()
````

## Zusatzaufgabe: Größte Setzung finden

Bei der Bauwerksüberwachung wurden an sechs Messpunkten die Setzungen (in mm) gemessen:

```python
setzungen = [2.4, 3.1, 2.9, 4.2, 3.8, 2.6]
```

Entwerfen Sie einen Algorithmus, der die **größte** Setzung findet, **ohne** die eingebaute Funktion `max()` zu verwenden. Notieren Sie ihn zuerst als **Pseudocode** (oder Flussdiagramm) und implementieren Sie ihn anschließend in Python.

Erwarteter Output:
```text
Größte gemessene Setzung: 4.2 mm
```

#### Lösungsvorschlag

Pseudocode:

```text
BEGINN
    LIES Liste s mit n Werten
    groesste = s[0]
    FÜR ALLE weiteren Werte s[i] der Liste:
        WENN s[i] > groesste DANN
            groesste = s[i]
    GIB groesste AUS
ENDE
```

Als Flussdiagramm entspricht der Algorithmus genau der Maximumsuche aus der Vorlesung: ein Schleifenkopf (orange Ellipse, »für alle Werte der Liste«) und darin eine blaue Raute (`s[i] > groesste ?`), die bei »wahr« den Merkwert aktualisiert.

````python
setzungen = [2.4, 3.1, 2.9, 4.2, 3.8, 2.6]   # mm

groesste = setzungen[0]
for i in range(1, len(setzungen)):
    if setzungen[i] > groesste:
        groesste = setzungen[i]

print("Größte gemessene Setzung:", groesste, "mm")
````
